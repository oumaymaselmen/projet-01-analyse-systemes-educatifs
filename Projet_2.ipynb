```python
# Exercice 2 :
  # Partie 1 : Réduire le périmètre en utilisant une approche métier
      # 1. Identification de la colonne des catégories métier

# -------------------------------------------------
# Import des packages
# -------------------------------------------------
import pandas as pd
import numpy as np

# Charger le fichier nettoyé
data_clean = pd.read_csv("EdStatsData_Clean.csv")

# Identifier la colonne métier
print("Identification de la colonne des catégories métier")
print("="*80)

print("\nRÉPONSE : La colonne qui décrit la catégorie métier des indicateurs est → 'Indicator Name'")

# Afficher quelques exemples
print("\nExemples de catégories d'indicateurs :")
print(data_clean[['Country Name', 'Indicator Code', 'Indicator Name']].head())

     # 2. Filtrage des catégories

#les catégories qui font sens par rapport à la demande de Mark et l’objectif du projet
# --- Étape 1 : Rechercher les catégories contenant  de mots-clés pertinents pour le projet

mots_cles = ['tertiary', 'students' , 'secondary']

# On filtre les catégories contenant au moins un mot-clé
categories_pertinentes  = data_clean['Indicator Name'].dropna().unique()
categories_selectionnees = [cat for cat in categories_pertinentes
                            if any(mot.lower() in cat.lower() for mot in mots_cles)]

print(" Catégories retenues via mots-clés :")
# Afficher juste les 3 premières catégories
for c in categories_selectionnees[:3]:
    print(" -", c)

      # 3. Afficher le nombre total d'indicateurs retenus
    
print(f"\nNombre total d'indicateurs après filtrage : {len(categories_selectionnees)}")

    # 4. Conserver seulement les colonnes pertinentes

# --- Étape 1 : Filtrer le dataframe selon ces catégories ---
new_data = data_clean[data_clean['Indicator Name'].isin(categories_selectionnees)].copy()

# --- Étape 2 :Filtrer les années pertinentes (2010-2015)

meta_cols = ['Country Name', 'Country Code', 'Indicator Name', 'Indicator Code']
annees = [str(y) for y in np.arange(2010, 2016)]
annees_disponibles = [y for y in annees if y in new_data.columns]

#  Sélection et affichage des années pertinentes pour l'analyse 

new_data = new_data[meta_cols + annees_disponibles].copy()
print(f"\nAnnées conservées pour l'analyse : {annees_disponibles}")
print(f"Nombre de lignes après filtrage : {len(new_data):,}")

print("Les colonnes 2010-2015 sont des données observées réelles.")
print("Les colonnes après 2015 sont des projections/futures et ont été exclues.")

# --- Étape 3 Calcul du taux de remplissage par indicateur

new_data['Taux Remplissage (%)'] = new_data[annees_disponibles].notnull().mean(axis=1) * 100

# Garder uniquement les indicateurs avec taux de remplissage > 80%
new_data = new_data[new_data['Taux Remplissage (%)'] > 80].copy()
print(f"Nombre d'indicateurs après filtrage (>80% remplissage) : {new_data['Indicator Name'].nunique()}")

      # 5.  Vérification des années futures

all_years_in_data = [col for col in data_clean.columns if col.isdigit()]
annees_futures = [col for col in all_years_in_data if int(col) > 2015]
print(f"\nAnnées futures identifiées et exclues : {annees_futures}")

# Exercice 2 :
  # Partie 2 : Réduire le périmètre en utilisant une approche Data
    # 1. Proportion d'indicateurs renseignés par année

proportion_indicateurs_par_annee = new_data[annees_disponibles].notnull().mean()
print("\nProportion d'indicateurs renseignés par année :")
print(proportion_indicateurs_par_annee)

     # 2. Proportion d'années renseignées par indicateur

proportion_annees_par_indicateur = new_data[annees_disponibles].notnull().mean(axis=1)
new_data['proportion_annees_renseignees'] = proportion_annees_par_indicateur
print("\nProportion d'années renseignées par indicateur :")
print(new_data[['Indicator Name', 'proportion_annees_renseignees']].sort_values(
    by='proportion_annees_renseignees', ascending=False).head(10))

     # 3. Réduction du nombre d'années et d'indicateurs

# Garder les années avec au moins 70% d'indicateurs renseignés
annees_pertinentes = proportion_indicateurs_par_annee[proportion_indicateurs_par_annee >= 0.7].index.tolist()
print(f"\nAnnées conservées pour la suite : {annees_pertinentes}")

# Garder les indicateurs avec au moins 70% d'années renseignées
indicateurs_pertinents = new_data[new_data['proportion_annees_renseignees'] >= 0.7].copy()
print(f"Nombre d'indicateurs conservés après filtrage : {indicateurs_pertinents['Indicator Name'].nunique()}")

# Nombre de pays avec valeur renseignée par indicateur et par année

pays_renseignes = indicateurs_pertinents.groupby('Indicator Name')[annees_pertinentes].apply(lambda x: x.notnull().sum())
pays_renseignes = pays_renseignes.reset_index()

# Mettre en format long pour trier facilement
pays_renseignes_long = pd.melt(pays_renseignes, id_vars='Indicator Name', var_name='Year', value_name='NbPaysRenseigne')
pays_renseignes_long = pays_renseignes_long.sort_values(by='NbPaysRenseigne', ascending=False)

print("\nExemple d'indicateurs les plus riches en données :")
print(pays_renseignes_long.head(10))

# Sélection finale d'une quinzaine d'indicateurs
#on choisit les 15 indicateurs les plus “complets” pour ne garder que ceux qui ont le plus de données disponibles sur les années et pays.
top_indicateurs = pays_renseignes_long.groupby('Indicator Name')['NbPaysRenseigne'].sum().sort_values(ascending=False).head(15)
liste_indicateurs_selectionnes = top_indicateurs.index.tolist()

print("\nListe finale des indicateurs sélectionnés (15 maximum) :")
print(liste_indicateurs_selectionnes)

# Exercice 2 :
  # Partie 3 : Consolider les résultats dans un dataframe (pays, indicateurs)
    # 1. Filtrer new_data pour ne garder que les indicateurs et années sélectionnés

final_data = new_data[new_data['Indicator Name'].isin(liste_indicateurs_selectionnes)].copy()
final_data = final_data[meta_cols + annees_pertinentes].copy()

    # 2. Transformer le dataframe : chaque ligne = pays, chaque colonne = indicateur (moyenne sur les années)

data_pivot = final_data.melt(id_vars=['Country Name', 'Country Code', 'Indicator Name'], 
                             value_vars=annees_pertinentes,
                             var_name='Year', value_name='Value')

# Pivot_table pour obtenir format pays x indicateur
data_finale = data_pivot.pivot_table(index=['Country Name', 'Country Code'],
                                     columns='Indicator Name',
                                     values='Value',
                                     aggfunc='mean').reset_index()

print("\nExemple de dataframe final agrégé :")
print(data_finale.head())


```

```python

# Exercice 3 
  # Partie 1 : Identifier les indicateurs redondants d'un point de vue statistique

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# Sélectionner uniquement les colonnes numériques (les indicateurs)
indicateurs_numeriques = data_finale.select_dtypes(include=np.number)

    # 1️. Calcul des matrices de corrélation

corr_pearson = indicateurs_numeriques.corr(method='pearson')
corr_spearman = indicateurs_numeriques.corr(method='spearman')

    # 2️. Visualisation avec heatmap

plt.figure(figsize=(10, 8))
sns.heatmap(corr_pearson, cmap='coolwarm', center=0)
plt.title("Matrice de corrélation (Pearson)")
plt.show()

plt.figure(figsize=(10, 8))
sns.heatmap(corr_spearman, cmap='coolwarm', center=0)
plt.title("Matrice de corrélation (Spearman)")
plt.show()

    # 3️. Suppression des indicateurs trop corrélés

# Seuil de corrélation absolue
seuil_corr = 0.7

# Trouver les colonnes redondantes
corr_matrix = corr_pearson.abs()
upper = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))

# Sélectionner les colonnes à supprimer
colonnes_a_supprimer = [column for column in upper.columns if any(upper[column] > seuil_corr)]

print(f"Indicateurs à supprimer (corrélation > {seuil_corr}):")
print(colonnes_a_supprimer)

# Créer un nouveau dataframe sans les indicateurs redondants
data_corr_reduit = data_finale.drop(columns=colonnes_a_supprimer)

print(f"\nNombre d'indicateurs avant : {data_finale.shape[1] - 2}")
print(f"Nombre d’indicateurs après suppression : {data_corr_reduit.shape[1] - 2}")

# Matrice finale nettoyée
print("\nAperçu du dataframe final non corrélé :")
display(data_corr_reduit.head())


# Exercice 3 :
  # Partie 2 : Analyser les indicateurs restants et formuler une liste de pays pertinents pour l'implantation d'Academy

# Import des librairies
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Charger le fichier final nettoyé
data_analysis = data_corr_reduit.copy()

# Choisir un indicateur pour analyse
indicateur = data_analysis.columns[4]  
print(f"Analyse de l'indicateur : {indicateur}")

# Statistiques descriptives
print("\nStatistiques descriptives :")
print(data_analysis[indicateur].describe())

# Distribution visuelle
plt.figure(figsize=(8, 5))
sns.displot(data_analysis[indicateur], kde=True)
plt.title(f"Distribution de l'indicateur : {indicateur}")
plt.show()

# Créer une fonction pour généraliser sur les autres indicateurs
def analyse_indicateur(df, indicateur):
    """
    Affiche les statistiques descriptives et la distribution
    d’un indicateur donné.
    """
    print(f"\n Indicateur : {'Population of the official age for lower secondary education, both sexes (number)'}")
    print(df['Population of the official age for lower secondary education, both sexes (number)'].describe())  # stats de base

    plt.figure(figsize=(8, 5))
    sns.displot(df[indicateur], kde=True)
    plt.title(f"Distribution : {indicateur}")
    plt.show()

# Appliquer à tous les indicateurs
colonnes_indicateurs = data_analysis.columns[2:]  # on ignore Country Name / Code

for col in colonnes_indicateurs:
    analyse_indicateur(data_analysis, col)

# Sélection quantitative des pays
# On prend le top 5 pays pour chaque indicateur
top_pays = {}
for col in colonnes_indicateurs:
    top_pays[col] = data_analysis.nlargest(5, col)['Country Name'].tolist()

# Trouver les pays qui reviennent souvent dans les top 5
tous_les_pays = sum(top_pays.values(), [])
pays_freq = pd.Series(tous_les_pays).value_counts()

# Garder ceux qui apparaissent au moins 2 fois 
pays_selectionnes = pays_freq[pays_freq >= 2].index.tolist()

print("\n Liste pertinente de pays selon les indicateurs :")
print(pays_selectionnes)
```
